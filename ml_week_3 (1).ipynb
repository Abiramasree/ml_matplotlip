```python
import pandas as pd
import numpy as np

data = pd.read_csv('final_vg.csv')

data.head()
```

```python
import matplotlib.pyplot as plt
```

```python
x_val = [0, 1, 2]
y_val = [3, 5, 9]

plt.plot(x_val, y_val)
plt.show()
```

```python
cat_count = data['Genre'].value_counts()

cat_count
```

```python
x_bar = cat_count.index
y_bar = cat_count

print(x_bar)
print(y_bar)
```

```python
plt.bar(x_bar, y_bar)

plt.show()
```

```python
plt.figure(figsize=(12, 8))

plt.bar(x_bar, y_bar)

plt.show()
```

```python
#Change the width of these bars
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2)
plt.xticks(rotation=90, fontsize=12)
```

```python
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.2, color='orange')
plt.xticks(rotation=90, fontsize=12)
plt.title("Games per Genre", fontsize= 15)
plt.xlabel("Genre")
plt.ylabel("Count of Games")
```
